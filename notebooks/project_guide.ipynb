# Kasvojen automaattinen hymyn tunnistus

Työn tarkoituksena on realiaikainen kasvojen tunnistus ja prosessointi, jonka perusteella arvataan hymyileekö tunnistuksen kohteena oleva henkilö vai ei.

Työn vaiheet
1. Kasvojen tunnistus
- MediaPipe kasvojen piirteet

2. Hymyn tunnistus
- MediaPipe Blendshapejen avulla hymyn tai mutrun tunnistus

## Käytetyt lähteet
https://ai.google.dev/edge/mediapipe/solutions/vision/face_landmarker/python

https://storage.googleapis.com/mediapipe-assets/Model%20Card%20Blendshape%20V2.pdf

Esitykseen kuuluu:
1. Valmiin työn esittely
2. Ryhmän jäsenten roolien kuvaukset
3. Prosessin kuvaus
4. Kuvaus projektityökalujen käytöstä
